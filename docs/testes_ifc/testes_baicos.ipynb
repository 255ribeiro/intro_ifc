```python
ifc = ifcopenshell.open("../ifc_data/teste_ifc_01.ifc")

def mostrar_estrutura(elemento, nivel=0):
    prefixo = "    " * nivel + ("└── " if nivel > 0 else "")
    nome = elemento.Name or "(sem nome)"
    print(f"{prefixo}[{elemento.is_a()}] {nome}")

    # Decomposição espacial (Projeto → Site → Edificio → Pavimento)
    for rel in getattr(elemento, "IsDecomposedBy", []):
        for filho in rel.RelatedObjects:
            mostrar_estrutura(filho, nivel + 1)

    # Elementos contidos no nível espacial
    for rel in getattr(elemento, "ContainsElements", []):
        tipos = {}
        for elem in rel.RelatedElements:
            t = elem.is_a()
            tipos[t] = tipos.get(t, 0) + 1
        for tipo, qtd in sorted(tipos.items()):
            print(f"{'    ' * (nivel + 1)}└── [{tipo}] × {qtd}")

for projeto in ifc.by_type("IfcProject"):
    mostrar_estrutura(projeto)
```
